```python
## Tipos de datos
##Aqui probamos numeros, texto y booleanos en Python.
```

```python
# Numeros
a = 10
b = 3.5

print("Entero:", a)
print("Decimal:", b)

# Texto
nombre = "Oscar"
print("Texto:", nombre)

#Booleanos
es_estudiante = True
print("Booleano:", es_estudiante)
```

```python
## Operaciones básicas
```

```python
# Operaciones aritméticas
print("Suma:", 5 + 3)
print("Resta:", 10 - 4)
print("Multiplicación:", 6*7)
print("División:", 15/2)
print("Módulo (residuo):", 15 % 2)

#Comparaciones
print("¿5 es mayor que 3?", 5 > 3)
print("¿10 es igual a 7?", 10 == 7)

# Operadores Lógicos
print("True AND False:", True and False)
print("True OR False:", True or False)
```

```python
## Listas y diccionarios
```

```python
# Lista
numeros = [1, 2, 3, 4, 5]
print("Lista:", numeros)
print("Primer elemento:", numeros[0])

# Diccionario
estudiante = {"nombre": "Ana", "edad": 12, "nota": 4.5}
print("Diccionario:", estudiante)
print("Edad:", estudiante["edad"])
```

```python
## Bucles y condicionales
```

```python
# Bucle for
for i in range(5):
    print("Iteración", i)

#Condicional
for n in numeros:
    print("Numero:", n)

# Condicional
if estudiante["nota"] >= 3:
    print(estudiante["nombre"], "aprobó")
else: print(estudiante["nombre"], "reprobó")
```
