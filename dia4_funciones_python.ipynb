```python
def saludar():
    print("Hola, bienvenido a Python!")

# Llamamos la función
saludar()
```

```python
def saludar_persona(nombre):
    print("Hola", nombre, "!")

saludar_persona("Oscar")
saludar_persona("Ana")
```

```python
def cuadrado(numero):
    return numero * numero

resultado = cuadrado(5)
print("El cuadrado de 5 es:", resultado)
def cuadrado(numero):
    return numero * numero

resultado = cuadrado(5)
print("El cuadrado de 5 es:", resultado)
```

```python
def promedio(notas):
    return sum(notas) / len(notas)

estudiante = {"nombre": "Ana", "notas": [3.5, 4.0, 4.5]}
prom = promedio(estudiante["notas"])

print("El promedio de", estudiante["nombre"], "es:", prom)
```
