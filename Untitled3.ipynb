```python
import pandas as pd

# Datos de ejemplo
data = {
    "nombre": ["Ana", "Luis", "Carla", "Pedro", "Sofia"],
    "nota1": [4.0, 3.5, 4.2, 3.0, 4.5],
    "nota2": [3.5, 4.0, 4.5, 2.8, 4.0],
    "nota3": [4.2, 3.8, 4.0, 3.5, 4.8]
}

# Crear DataFrame
df = pd.DataFrame(data)

# Guardar CSV
df.to_csv("estudiantes.csv", index=False)

# Verificar contenido
df
```

```python
# Cargar CSV
df = pd.read_csv("estudiantes.csv")

# Mostrar primeras filas
df.head()
```

```python
# Calcular promedio por estudiante
df['promedio'] = df[['nota1','nota2','nota3']].mean(axis=1)

# Mostrar resultados
df
```

```python
promedio_grupo = df['promedio'].mean()
print(f"Promedio general del grupo: {round(promedio_grupo, 2)}")
```

```python
import matplotlib.pyplot as plt

# Crear gráfico de barras
plt.bar(df['nombre'], df['promedio'], color='skyblue')
plt.axhline(promedio_grupo, color='red', linestyle='--', label='Promedio grupo')
plt.ylabel("Promedio")
plt.title("Promedio por estudiante")
plt.legend()
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Crear sliders para cada estudiante
sliders = {}
for nombre in df['nombre']:
    sliders[nombre] = widgets.FloatSlider(
        value=df.loc[df['nombre'] == nombre, 'promedio'].values[0],
        min=0, max=5, step=0.1,
        description=nombre
    )

# Función para actualizar gráfico
def actualizar_dashboard(**kwargs):
    clear_output(wait=True)
    
    # Actualizar promedios del DataFrame
    for nombre, valor in kwargs.items():
        df.loc[df['nombre'] == nombre, 'promedio'] = valor
    
    promedio_grupo = df['promedio'].mean()
    
    # Mostrar tabla
    display(df)
    
    # Gráfico
    fig, ax = plt.subplots(figsize=(8,4))
    ax.bar(df['nombre'], df['promedio'], color='skyblue')
    ax.axhline(promedio_grupo, color='red', linestyle='--', label='Promedio grupo')
    ax.set_ylabel("Promedio")
    ax.set_title("Promedio por estudiante (Interactivo)")
    ax.legend()
    plt.show()

# Crear interfaz interactiva
ui = widgets.VBox(list(sliders.values()))
out = widgets.interactive_output(actualizar_dashboard, sliders)

display(ui, out)
```

```python
# Función para asignar color según promedio
def color_por_promedio(valor):
    if valor >= 4.0:
        return 'green'      # Excelente
    elif valor >= 3.0:
        return 'yellow'     # Promedio
    else:
        return 'red'        # Bajo

# Función para actualizar dashboard con colores
def actualizar_dashboard_colores(**kwargs):
    clear_output(wait=True)
    
    # Actualizar promedios del DataFrame
    for nombre, valor in kwargs.items():
        df.loc[df['nombre'] == nombre, 'promedio'] = valor
    
    promedio_grupo = df['promedio'].mean()
    
    # Mostrar tabla
    display(df)
    
    # Colores para las barras
    colores = [color_por_promedio(x) for x in df['promedio']]
    
    # Gráfico
    fig, ax = plt.subplots(figsize=(8,4))
    ax.bar(df['nombre'], df['promedio'], color=colores)
    ax.axhline(promedio_grupo, color='red', linestyle='--', label='Promedio grupo')
    ax.set_ylabel("Promedio")
    ax.set_title("Promedio por estudiante (Interactivo y Coloreado)")
    ax.legend()
    plt.show()

# Crear interfaz interactiva
ui = widgets.VBox(list(sliders.values()))
out = widgets.interactive_output(actualizar_dashboard_colores, sliders)

display(ui, out)
```

```python
# Función para generar mensaje según promedio
def mensaje_pedagogico(valor):
    if valor >= 4.0:
        return "¡Excelente desempeño! Sigue así 💪"
    elif valor >= 3.0:
        return "Buen desempeño, pero hay espacio para mejorar 🙂"
    else:
        return "Necesita reforzar conocimientos 🔴"

# Función para actualizar dashboard con colores y mensajes
def actualizar_dashboard_mensajes(**kwargs):
    clear_output(wait=True)
    
    # Actualizar promedios del DataFrame
    for nombre, valor in kwargs.items():
        df.loc[df['nombre'] == nombre, 'promedio'] = valor
    
    promedio_grupo = df['promedio'].mean()
    
    # Mostrar tabla con promedio actualizado
    display(df)
    
    # Colores para las barras
    colores = [color_por_promedio(x) for x in df['promedio']]
    
    # Crear gráfico
    fig, ax = plt.subplots(figsize=(8,4))
    ax.bar(df['nombre'], df['promedio'], color=colores)
    ax.axhline(promedio_grupo, color='red', linestyle='--', label='Promedio grupo')
    ax.set_ylabel("Promedio")
    ax.set_title("Promedio por estudiante (Interactivo, Coloreado y Mensajes)")
    ax.legend()
    plt.show()
    
    # Mostrar mensajes pedagógicos
    print("📚 Mensajes pedagógicos por estudiante:")
    for nombre, valor in kwargs.items():
        print(f"{nombre}: {mensaje_pedagogico(valor)}")

# Interfaz interactiva
ui = widgets.VBox(list(sliders.values()))
out = widgets.interactive_output(actualizar_dashboard_mensajes, sliders)

display(ui, out)
```

```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt

# Datos iniciales
data = {
    "nombre": ["Ana", "Luis", "Carla", "Pedro", "Sofia"],
    "nota1": [4.0, 3.5, 4.2, 3.0, 4.5],
    "nota2": [3.5, 4.0, 4.5, 2.8, 4.0],
    "nota3": [4.2, 3.8, 4.0, 3.5, 4.8]
}

df = pd.DataFrame(data)

# Funciones
def color_por_promedio(valor):
    if valor >= 4.0:
        return 'green'
    elif valor >= 3.0:
        return 'yellow'
    else:
        return 'red'

def mensaje_pedagogico(valor):
    if valor >= 4.0:
        return "¡Excelente desempeño! Sigue así 💪"
    elif valor >= 3.0:
        return "Buen desempeño, pero hay espacio para mejorar 🙂"
    else:
        return "Necesita reforzar conocimientos 🔴"

# Título
st.title("Mini Dashboard Educativo Interactivo")

# Sliders para cada estudiante
for nombre in df['nombre']:
    df.loc[df['nombre'] == nombre, 'promedio'] = st.slider(
        nombre,
        min_value=0.0,
        max_value=5.0,
        value=float(df.loc[df['nombre'] == nombre, ['nota1','nota2','nota3']].mean()),
        step=0.1
    )

# Promedio general
promedio_grupo = df['promedio'].mean()
st.write(f"**Promedio general del grupo:** {round(promedio_grupo,2)}")

# Gráfico de barras
colores = [color_por_promedio(x) for x in df['promedio']]
fig, ax = plt.subplots(figsize=(8,4))
ax.bar(df['nombre'], df['promedio'], color=colores)
ax.axhline(promedio_grupo, color='red', linestyle='--', label='Promedio grupo')
ax.set_ylabel("Promedio")
ax.set_title("Promedio por estudiante")
ax.legend()
st.pyplot(fig)

# Mensajes pedagógicos
st.write("📚 **Mensajes pedagógicos por estudiante:**")
for nombre, valor in zip(df['nombre'], df['promedio']):
    st.write(f"{nombre}: {mensaje_pedagogico(valor)}")
```
