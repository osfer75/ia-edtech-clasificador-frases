```python
estudiantes = ["Ana", "Luis", "Marta"]
print(estudiantes[0])   # Primer estudiante
estudiantes.append("Carlos")  # Agregar un nuevo estudiante
print(estudiantes)
```

```python
estudiante = {"nombre": "Ana", "edad": 12, "notas": [4.0, 3.5, 4.5]}
print(estudiante["nombre"])
print(estudiante["notas"])
```

```python
grupo = [
    {"nombre": "Ana", "notas": [4.0, 3.5, 4.5]},
    {"nombre": "Luis", "notas": [3.0, 4.0, 3.5]},
    {"nombre": "Marta", "notas": [5.0, 4.5, 4.0]}
]

for est in grupo:
    prom = sum(est["notas"]) / len(est["notas"])
    print("El promedio de", est["nombre"], "es", prom)
```
